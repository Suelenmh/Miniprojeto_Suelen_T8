```python
'''ler a base'''
import pandas as pd

df = pd.read_csv("Base Varejo.csv", sep=";")
df_original = df.copy()   # cópia para comparar antes/depois
print("Linhas e colunas:", df.shape)
df.head()
```

```python
'''colunas e tipos'''
print("Colunas:", list(df.columns))
df.info()
```

```python
'''valores ausente'''
print(df.isna().sum())
print((df.isna().mean() * 100).round(2))
```

```python
'''textos vazios ou só espaços'''
colunas_texto = ["DATA", "CL_GENERO", "CL_SEG", "PR_CAT", "PR_NOME"]
for c in colunas_texto:
    qtd = df[c].astype("string").str.strip().eq("").sum()
    print(c, "->", qtd)
```

```python
'''duplicatas'''
print("Linhas duplicadas:", df.duplicated().sum())
df[df.duplicated(keep=False)].head(10)
```

```python
'''inconsistências de texto'''
print(df["PR_CAT"].value_counts(dropna=False))
print("PR_ID com #N/D:", df[df["PR_CAT"] == "#N/D"]["PR_ID"].unique())
```

```python
'''nomes de produto fora do padrão (com letra minúscula)'''
com_minuscula = df[df["PR_NOME"] != df["PR_NOME"].str.upper()]
print("Linhas com PR_NOME fora do padrão:", len(com_minuscula))
print(com_minuscula["PR_NOME"].value_counts())
```

```python
print("Compras diferentes (CO_ID):", df["CO_ID"].nunique())
print("Clientes diferentes (CL_ID):", df["CL_ID"].nunique())
```
