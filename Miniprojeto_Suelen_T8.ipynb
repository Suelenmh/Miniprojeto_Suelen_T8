# Sprint 2 e 3 - Limpeza, padronização e funções

```python
'''função 1: tratar os dados'''
def tratar_dados(tabela):
    """Recebe a tabela bruta e devolve a tabela tratada."""
    tabela = tabela.copy()

    # 1) remove colunas 100% vazias
    tabela = tabela.dropna(axis=1, how="all")

    # 2) padroniza texto (sem espaços nas pontas, tudo MAIÚSCULO)
    #    e troca "#N/D" por "NAO INFORMADO" (sem apagar a linha)
    for coluna in ["PR_CAT", "PR_NOME"]:
        tabela[coluna] = tabela[coluna].str.strip().str.upper()
        tabela[coluna] = tabela[coluna].replace("#N/D", "NAO INFORMADO")

    # 3) transforma DATA em data de verdade
    tabela["DATA"] = pd.to_datetime(tabela["DATA"], format="%d/%m/%Y")

    return tabela
```

```python
'''função 2: resumo de uma coluna numérica'''
def resumo_numerico(serie):
    """Recebe uma coluna numérica e devolve um resumo."""
    return {
        "contagem": int(serie.count()),
        "media": round(float(serie.mean()), 2),
        "minimo": serie.min(),
        "maximo": serie.max(),
    }
```

```python
'''aplicar a limpeza'''
df_limpo = tratar_dados(df_original)
print("Registros antes :", len(df_original), "| colunas:", df_original.shape[1])
print("Registros depois:", len(df_limpo), "| colunas:", df_limpo.shape[1])
```

```python
'''comparação antes e depois'''
print("ANTES")
print(df_original["PR_CAT"].value_counts(dropna=False))
print("\nDEPOIS")
print(df_limpo["PR_CAT"].value_counts(dropna=False))

print("\nREFRIGERANTE LIMaO -> antes:", (df_original["PR_NOME"] == "REFRIGERANTE LIMaO").sum())
print("REFRIGERANTE LIMaO -> depois:", (df_limpo["PR_NOME"] == "REFRIGERANTE LIMaO").sum())
print("REFRIGERANTE LIMAO -> depois:", (df_limpo["PR_NOME"] == "REFRIGERANTE LIMAO").sum())

print("\nNomes NAO INFORMADO:", (df_limpo["PR_NOME"] == "NAO INFORMADO").sum())
print("Duplicatas mantidas:", df_limpo.duplicated().sum())
print("Nulos no total:", df_limpo.isna().sum().sum())
print("Tipo da DATA:", df_limpo["DATA"].dtype)
```

## Decisões de limpeza

Antes de limpar, fiz o diagnóstico da base e só tratei o que realmente era problema. Cada decisão está justificada abaixo.

| Problema encontrado | Decisão | Motivo |
|---|---|---|
| 4 colunas `Unnamed` 100% vazias (830.000 ausentes cada) | **Removidas** | Não têm nenhuma informação. Vieram de `;` sobrando no final das linhas do CSV. |
| Valores ausentes nas outras 10 colunas | **Nada a tratar** | O diagnóstico encontrou 0 ausentes, então não preenchi nem removi nada. |
| Textos vazios ou só espaços | **Nada a tratar** | O diagnóstico encontrou 0 em todas as colunas de texto. |
| 96.553 linhas duplicadas | **Mantidas** | A base não tem coluna de quantidade. Linhas iguais dentro da mesma compra provavelmente são o mesmo produto levado mais de uma vez, e não consigo confirmar que sejam erro. Removê-las poderia apagar compras reais. |
| `#N/D` em `PR_CAT` e `PR_NOME` (3.650 linhas, todas do produto 107) | **Trocado por `NAO INFORMADO`** | São compras reais, então não removi as linhas. Também não inventei uma categoria que a base não informa. |
| `REFRIGERANTE LIMaO` com letra minúscula (7.320 linhas) | **Padronizado para maiúsculas** | Corrige a inconsistência sem mudar o significado do dado. |
| `DATA` armazenada como texto | **Convertida para data** | Permite ordenar e filtrar por período corretamente. |

### Comparação antes e depois

- **Registros:** 830.000 antes e 830.000 depois (nenhuma linha foi apagada).
- **Colunas:** 14 antes e 10 depois (removi as 4 vazias).
- **`#N/D`:** de 3.650 para 0, que passaram a `NAO INFORMADO` (3.650).
- **`REFRIGERANTE LIMaO`:** de 7.320 para 0, que passaram a `REFRIGERANTE LIMAO` (7.320).
- **Duplicatas mantidas:** 96.553. **Nulos no total:** 0.
- **Tipo de `DATA`:** de texto para `datetime64`.

### Efeito na qualidade da análise

Como preservei todas as linhas, as contagens por categoria e por compra continuam representando o que aconteceu na loja. Padronizar o texto evita que o mesmo produto seja contado como dois itens diferentes, e marcar `NAO INFORMADO` mantém essas compras visíveis nos resultados, sem inventar informação.
